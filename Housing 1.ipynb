# 13. CORRELATION HEATMAP

```python
numeric = Housing.select_dtypes(include=np.number)

plt.figure(figsize=(10,8))

sns.heatmap(numeric.corr(),annot=True,cmap="coolwarm",fmt=".2f", linewidths=0.5)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.savefig("images/26_heatmap.png", dpi=300)

plt.show()
```

# 14. PAIRPLOT

# 15. CORRELATION WITH PRICE

```python

corr = numeric.corr()["price"].sort_values(ascending=False)

print("\nCorrelation with Price")
print(corr)

plt.figure(figsize=(8,6))

corr.drop("price").plot(kind="barh")

plt.title("Factors Influencing House Prices")

plt.xlabel("Correlation")

plt.tight_layout()

plt.savefig("images/28_price_correlation.png", dpi=300)

plt.show()
```
